```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips=sns.load_dataset("tips")
print(tips)
sns.set_style('whitegrid')
tipgraph= sns.lineplot(data=tips, x='total_bill', y='tip',color = 'hotpink',
    linestyle = '--')
tipgraph.set(title='Tips vs. total bill',
             xlabel='Total bill',
             ylabel='total tip')
```

```python
sns.set_style('whitegrid')
tipgraph2= sns.scatterplot(data=tips, x='total_bill', y='tip',style = 'time', hue = 'day', palette=['purple','hotpink','deepskyblue','yellowgreen'])
tipgraph2.set(title='Tips vs. total bill',
             xlabel='Total bill',
             ylabel='total tip')
```

```python
sns.pairplot(
    data = tips, 
    hue = 'day'
)
```
