```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(['Luffy', 'Zoro','Nami','Usopp','Sanji'])
y2 = np.array([110,180,240,99,220])

fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1,figsize=(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)

ax1.set_title('Title for graph 1')
ax2.set_title('Title for graph 2')

ax1.set_xlabel('Numbers')
ax2.set_xlabel('Names')

fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],['ax2','ax3']],figsize=(10,5))
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
```

```python
x = np.array(['Luffy', 'Zoro','Nami','Usopp','Sanji'])
y1=np.array([110,280,240,99,220])
y2=np.array([170,100,90,120,50])

fig,ax=plt.subplots(figsize=(7,3))
ax.bar(x, y1, color='indigo')
ax.plot(x,y2,color='red')
y2_sd=np.std(y2)
ax.errorbar(x,
            y2,
            yerr = y2_sd,
            fmt = 'none',
            ecolor ='green',
            elinewidth=3,
            capsize=6,
            capthick=4,
            errorevery=2) #err bar every 2 data points
```
