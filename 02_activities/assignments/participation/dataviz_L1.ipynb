```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

Axes --> the actual plot or graph (not the x and y axes)


```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y)

fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1={'family':'sans serif', 'color': 'blue', 'size': 20}
font2={'family':'monospace', 'color': 'green', 'size': 14}
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, linestyle='--', linewidth=1, color='darkorchid')
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start')
fig.tight_layout()
ax.grid(axis = 'y')

ax.scatter(x,
           y,
           marker='x',
           color='red',
           )
```
