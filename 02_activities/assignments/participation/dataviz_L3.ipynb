```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(x,y1)
ax.plot(x,y2)
ax.plot(x,y1,label = "Person 1")
ax.plot(x,y2,label = "Person 2")
ax.legend (loc='upper right', #lower, upper; L/R
           frameon = True,#False
           fontsize = 12,
           ncol = 1,
           shadow = False) #False
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(x,y1)
ax.plot(x,y2)
ax.plot(x,y1,label = "Person 1")
ax.plot(x,y2,label = "Person 2")
ax.legend (loc='upper left', #lower, upper; L/R
           frameon = True,#False
           fontsize = 12,
           ncol = 1,
           shadow = False,
           bbox_to_anchor=(1,1)) #False

ax.text(10,95,"Important value")
```

```python
fig, ax = plt.subplots()
ax.axis([0,10,0,10])

ax.text(1,5,". Data: (1,5)",transform=ax.transData)
ax.text(0.5, 0.1, ". Axes:(0.5,0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure:(0.2,0.2)", transform=fig.transFigure)
```

```python

fig, ax = plt.subplots(figsize=(8,4))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2, label = 'person 2')
ax.legend(loc='lower right')

ax.annotate('FYI!', xy=(10,95),xytext=(20,94), arrowprops=dict(facecolor='black'))
```

```python

fig, ax = plt.subplots(figsize=(8,4))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2, label = 'person 2')
ax.legend(loc='lower right')

ax.annotate('FYI!', xy=(10,95),xytext=(20,94), arrowprops=dict(arrowstyle='wedge',facecolor='black'))
```

```python

fig, ax = plt.subplots(figsize=(8,4))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2, label = 'person 2')
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator()) #does not retain the tick marks
ax.xaxis.set_major_formatter(plt.NullFormatter()) #retains the tick marks

ax.annotate('FYI!', xy=(10,95),xytext=(20,94), arrowprops=dict(arrowstyle='wedge',facecolor='black'))
```

```python

fig, ax = plt.subplots(figsize=(8,4))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2, label = 'person 2')
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(5)) #5 tick marks only
plt.xticks(rotation = 90, ha='right')
ax.set_xlabel('X axis title', color='indigo')
ax.set_title('Title goes here')
ax.annotate('FYI!', xy=(10,95),xytext=(20,94), arrowprops=dict(arrowstyle='wedge',facecolor='black'))
```
